```python
# Part A

# Create a Pandas Series from a list of temperatures measured in Celsius

# Part B

# Convert the temperatures to Fahrenheit and add them as a new Series with the same index
```

```python
import pandas as pd
celsius = [25, 30, 15, 20, 35]
temp_c = pd.Series(celsius, name="Celsius")
print("Temperatures in Celsius:\n", temp_c)
```

```python
# Part B: Convert Celsius to Fahrenheit and create a new Series
temp_f = temp_c * 9/5 + 32
temp_f.name = "Fahrenheit"
print("\nTemperatures in Fahrenheit:\n", temp_f)
```
