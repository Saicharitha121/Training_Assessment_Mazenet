```python
 # Matrix Creation

# Using NumPy, create a 5×5 matrix with values ranging from 1 to 25

# Diagonal Extraction

# Extract the diagonal elements from your generated matrix

# Calculation

# Calculate the sum of the extracted diagonal elements
```

```python
import numpy as np
matrix = np.arange(1, 26).reshape(5, 5)
print("5x5 Matrix:\n", matrix)
```

```python
diagonal = np.diag(matrix)
print("\nDiagonal Elements:", diagonal)
```

```python
sum_diagonal = np.sum(diagonal)
print("\nSum of Diagonal Elements:", sum_diagonal)
```
