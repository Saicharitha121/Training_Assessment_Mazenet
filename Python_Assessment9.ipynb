```python
# # Data Preparation

# Start with a sales DataFrame containing product categories and sales figures

# Aggregation

# Calculate total sales per product category

# Visualisation

# Using Matplotlib or Pandas plotting, create a bar chart displaying the results
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Category': ['pen', 'pencil', 'Notebooks', 'Erasers', 'scales'],
    'Sales': [15000, 8000, 12000, 18000, 7000]
}
df = pd.DataFrame(data)
print("Sales Data:\n", df)
```

```python
total_sales = df.groupby('Category')['Sales'].sum()
print("\nTotal Sales per Category:\n", total_sales)
```

```python

total_sales.plot(kind='bar', color='skyblue')
plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.show()
```
