```python
# Given a list of words, create a dictionary where keys are words and values are the counts of vowels in each word.
```

```python
words = input("Enter words separated by space: ").split()
vowels = 'aeiouAEIOU'
vowel_count = {}

for word in words:
    count = sum(1 for ch in word if ch in vowels)
    vowel_count[word] = count

print("Dictionary with vowel counts:", vowel_count)
```
